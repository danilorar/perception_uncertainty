### Problem statement

The built-in esmini [controllers](https://esmini.github.io/controllers.html), such as `ACCController`, execute inside the esmini C++ process and consume detections from the ideal `ObjectSensor` through the perception interface. 

### Perception architecture

The perception pipeline is layered:

```bash
OpenSCENARIO scenario
        |
        v
esmini ground-truth object state
        |
        v
ObjectSensor / IdealSensor
        |
        |  geometric detection: range, FOV, visibility, relative state
        v
PerceptionModel 
        |  
        v
Perceived detections
        |
        v
Built-in C++ controller, for example ACCController
        |
        v
Ego-vehicle control
```

The `ObjectSensor` is the ideal geometric sensor and it determines whether an object is visible to the sensor and calculates its ideal relative measurements.

 The `PerceptionModel` is applied on top of that ideal result. In noise-free mode, the perceived measurement equals the ideal measurement. In noisy mode, the model modifies the measurement according to the configured uncertainty distribution.

The physical ground truth is never overwritten. Instead, each detection carries separate values:

```bash
true_distance       ground-truth measurement from esmini
perceived_distance  measurement provided to the controller
distance_error      perceived_distance - true_distance
```

This separation is essential for evaluation. The controller must use `perceived_distance`, while the evaluator can use `true_distance` to measure how perception quality affected the controller.

````python
from pathlib import Path
import subprocess

project_root = Path("/home/danilo/Desktop/perception-models")

esmini = (
    project_root
    / "esmini"
    / "build"
    / "EnvironmentSimulator"
    / "Applications"
    / "esmini"
    / "esmini"
)

scenarios = {
    "ideal": project_root / "OSC-NCAP-scenarios" / "OpenSCENARIO" / "NCAP" / "AEB_C2C_2023" / "NCAP_AEB_C2C_CCFhol_2023_r157_ideal.xosc",
    "noisy": project_root / "OSC-NCAP-scenarios" / "OpenSCENARIO" / "NCAP" / "AEB_C2C_2023" / "NCAP_AEB_C2C_CCFhol_2023_r157_noisy.xosc",
}

if not esmini.exists():
    raise FileNotFoundError(f"esmini executable not found: {esmini}")

for scenario in scenarios.values():
    if not scenario.exists():
        raise FileNotFoundError(f"scenario file not found: {scenario}")
````

````python
for mode, scenario in scenarios.items():
    print(f"Running {mode}: {scenario.name}", flush=True)

    command = [
        str(esmini),
        "--window", "60", "60", "1200", "700",
        "--osc", str(scenario),
    ]

    subprocess.run(
        command,
        cwd=project_root,
        check=True,
    )

    print(f"{mode} scenario completed.", flush=True)
````
